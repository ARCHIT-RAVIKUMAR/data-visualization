# Spotify Tracks Dataset - Exploratory Data Analysis & Data Visualization

## Objective

The objective of this notebook is to explore the Spotify Tracks dataset, perform exploratory data analysis (EDA), create meaningful visualizations using Matplotlib and Seaborn, and communicate insights through effective data storytelling.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
```

## Load Dataset

```python
df = pd.read_csv("dataset.csv")

df.head()
```

## Dataset Overview

```python
print("Dataset Shape:", df.shape)

df.info()
```

```python
df.describe()
```

```python
df.describe(include="object")
```

```python
df.tail()
```

## Missing Values & Duplicate Records

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

## Numerical and Categorical Features

```python
numerical_features = df.select_dtypes(include=["int64", "float64"]).columns
categorical_features = df.select_dtypes(include=["object"]).columns

print("Numerical Features")
print(numerical_features)

print("\nCategorical Features")
print(categorical_features)
```

```python
for col in categorical_features:
    print(f"\n{col}")
    print(df[col].nunique(), "unique values")
```

```python
df.columns
```

## Visualization 1: Distribution of Track Popularity

```python
plt.figure(figsize=(10,6))
plt.hist(df["popularity"], bins=20)
plt.title("Distribution of Spotify Track Popularity")
plt.xlabel("Popularity")
plt.ylabel("Number of Tracks")

plt.savefig("visualization1.png", dpi=300, bbox_inches="tight")

plt.show()
```

**Insight:** Most Spotify tracks have moderate popularity, while only a small number achieve very high popularity.

## Visualization 2: Top 10 Most Common Music Genres

```python
top_genres = df["track_genre"].value_counts().head(10)

plt.figure(figsize=(12,6))
top_genres.plot(kind="bar")

plt.title("Top 10 Music Genres")
plt.xlabel("Genre")
plt.ylabel("Number of Tracks")
plt.xticks(rotation=45)

plt.savefig("visualization2.png", dpi=300, bbox_inches="tight")

plt.show()
```

**Insight:** A few genres dominate the dataset, indicating that Spotify contains significantly more tracks from popular genres.

## Visualization 3: Explicit Content Distribution

```python
explicit_counts = df["explicit"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    explicit_counts,
    labels=["Non-Explicit", "Explicit"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Explicit vs Non-Explicit Tracks")

plt.savefig("visualization3.png", dpi=300, bbox_inches="tight")

plt.show()
```

**Insight:** The majority of tracks are non-explicit, while explicit tracks represent a smaller portion of the dataset.

## Visualization 4: Danceability vs Energy

```python
plt.figure(figsize=(8,6))
plt.scatter(df["danceability"], df["energy"], alpha=0.4)

plt.title("Danceability vs Energy")
plt.xlabel("Danceability")
plt.ylabel("Energy")

plt.savefig("visualization4.png", dpi=300, bbox_inches="tight")

plt.show()
```

**Insight:** Tracks with higher danceability often exhibit higher energy levels, although there is noticeable variation.

## Visualization 5: Correlation Between Audio Features

```python
plt.figure(figsize=(12,8))

sns.heatmap(
    df[numerical_features].corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.savefig("visualization5.png", dpi=300, bbox_inches="tight")

plt.show()
```

**Insight:** Some audio features exhibit moderate correlations, while most remain relatively independent of one another.

## Visualization 6: Popularity by Explicit Content

```python
plt.figure(figsize=(8,6))

sns.boxplot(
    x="explicit",
    y="popularity",
    data=df
)

plt.title("Popularity by Explicit Content")

plt.savefig("visualization6.png", dpi=300, bbox_inches="tight")

plt.show()
```

**Insight:** Explicit and non-explicit tracks show overlapping popularity ranges, suggesting that explicit content alone does not determine popularity.

# Overall Conclusion

The Spotify Tracks dataset reveals diverse musical characteristics and listening trends. While a few genres dominate the dataset, popularity is influenced by multiple audio features rather than a single characteristic. Relationships between danceability, energy, and other musical attributes provide valuable insights for recommendation systems and future music analytics.
